# Python：成績排序與及格邊界

這份筆記整理一個常見的成績處理問題：

給定一組成績，以 **60 分** 為及格標準，需要：

1. 將所有成績由小到大排序
2. 找出 **最高不及格分數**
3. 找出 **最低及格分數**
4. 處理「全部及格」或「全部不及格」的特殊情況

過程中會練習：

- `input().split()`
- 字串轉整數
- list 的參照與複製
- `sort()` / `sorted()`
- `for`、`if`、`break`
- list comprehension


## 1. 讀入一整行成績

假設輸入：

```text
0 11 22 33 55 66 77 99 88 44
```

使用：

```python
scores = input().split()
```

會得到：

```python
['0', '11', '22', '33', '55', '66', '77', '99', '88', '44']
```

注意：這些資料目前仍然是 **字串**。


````python
scores = input("請輸入成績：").split()

print(scores)
print(type(scores[0]))
````

## 2. 把每一個成績轉成整數

不能直接寫：

```python
int(['0', '11', '22'])
```

因為 `int()` 只能把 **單一字串或數值** 轉成整數，不能一次轉換整個 list。

可以使用迴圈逐一轉換：


````python
scores = input("請輸入成績：").split()

for i in range(len(scores)):
    scores[i] = int(scores[i])

print(scores)
````

更常見的簡潔寫法是：

```python
scores = list(map(int, input().split()))
```

執行順序：

```text
input()
   ↓
"0 11 22 33"
   ↓ split()
["0", "11", "22", "33"]
   ↓ map(int, ...)
0, 11, 22, 33
   ↓ list(...)
[0, 11, 22, 33]
```


````python
scores = list(map(int, input("請輸入成績：").split()))

print(scores)
````

# 3. List 的「參照」與「複製」

這兩種寫法看起來很像，但意思不同。

### 直接指定

```python
lst_2 = lst
```

此時 `lst` 和 `lst_2` 指向 **同一個 list**。

因此修改其中一個，另一個也會看到相同變化。


````python
lst = [0, 11, 22, 33, 55, 66, 77, 99, 88, 44]

lst_2 = lst

lst.sort()

print("lst   =", lst)
print("lst_2 =", lst_2)
````

如果需要一份獨立的 list，可以使用切片：

```python
lst_2 = lst[:]
```

此時 `lst_2` 是一份新的 list。

例如往 `lst_2` 加入 `60`，不會影響原本的 `lst`。


````python
lst = [0, 11, 22, 33, 55, 66, 77, 99, 88, 44]

lst_2 = lst[:]
lst_2.append(60)

lst.sort()
lst_2.sort()

print("lst   =", lst)
print("lst_2 =", lst_2)
````

# 4. 先將成績排序

要找「最高不及格」與「最低及格」，先排序會讓問題簡單很多。

```python
scores.sort()
```

會直接修改原本的 list。

而：

```python
sorted(scores)
```

會建立一個新的排序結果。


````python
scores = [0, 11, 22, 33, 55, 66, 77, 99, 88, 44]

scores.sort()

print(scores)
````

排序後：

```text
0 11 22 33 44 55 66 77 88 99
```

以 `60` 分為界：

```text
0 11 22 33 44 55 | 66 77 88 99
                 ↑
               60 分
```

因此：

- 最高不及格：`55`
- 最低及格：`66`


# 5. 用迴圈找及格邊界

排序完成後，可以從左到右掃描。

當：

```python
score < 60
```

就持續更新「最高不及格」。

第一次遇到：

```python
score >= 60
```

就知道它是「最低及格」，因此可以 `break`。


````python
scores = [0, 11, 22, 33, 44, 55, 66, 77, 88, 99]

low = None
high = None

for score in scores:
    if score < 60:
        low = score
    else:
        high = score
        break

print("最高不及格：", low)
print("最低及格：", high)
````

## 為什麼遇到第一個及格分數就可以 `break`？

因為成績已經由小到大排序。

例如：

```text
0 11 22 33 44 55 66 77 88 99
```

第一個 `>= 60` 的分數一定就是 **最低及格分數**。

後面的成績只會更高，因此不需要繼續搜尋。

# 7. 完整寫法：使用迴圈

假設輸入格式為：

```text
n
n 個成績
```

例如：

```text
10
0 11 22 33 55 66 77 99 88 44
```


````python
n = int(input())
scores = list(map(int, input().split()))

scores.sort()

print(*scores)

low = "best case"
high = "worst case"

for score in scores:
    if score < 60:
        low = score
    else:
        high = score
        break

print(low)
print(high)
````

這裡：

```python
print(*scores)
```

會把 list 中的元素展開輸出。

例如：

```python
scores = [0, 11, 22]
print(*scores)
```

輸出：

```text
0 11 22
```


# 8. 更簡潔的寫法：List Comprehension

可以先把成績分成兩組：

```python
fail = [x for x in scores if x < 60]
passed = [x for x in scores if x >= 60]
```

因為 `scores` 已經排序：

- `fail[-1]`：最高不及格
- `passed[0]`：最低及格


````python
n = int(input())
scores = sorted(map(int, input().split()))

print(*scores)

fail = [x for x in scores if x < 60]
passed = [x for x in scores if x >= 60]

print(fail[-1] if fail else "best case")
print(passed[0] if passed else "worst case")
````

## 條件運算式

這一行：

```python
fail[-1] if fail else "best case"
```

意思是：

```python
if fail:
    print(fail[-1])
else:
    print("best case")
```

同理：

```python
passed[0] if passed else "worst case"
```

表示：

- 如果 `passed` 不是空的 → 取第一個元素
- 如果 `passed` 是空的 → `"worst case"`

# 10. 補充：`min()` 與 `max()`

原本筆記最後也練習了：

```python
min(lst)
max(lst)
```

例如：


````python
scores = list(map(int, input("請輸入數個分數：").split()))

lowest = min(scores)
highest = max(scores)
score_range = highest - lowest

print("最低分：", lowest)
print("最高分：", highest)
print("分數範圍：", score_range)
````

例如輸入：

```text
0 11 22 33 55 66 77 99 88 44
```

則：

```text
最低分 = 0
最高分 = 99
分數範圍 = 99
```

這和前面的「及格邊界」是不同問題：

- `min()` / `max()`：找整體極值
- 及格邊界：找 `60` 左右最接近的兩個值
