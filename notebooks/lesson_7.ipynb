## 1. `for` 迴圈與迴圈變數

原始程式：

````python
for i in range(3):  # i 依序取得 0、1、2
    print(i)
    i = 3

print(i)
````

### 重點說明

`range(3)` 會依序產生：

```text
0, 1, 2
```

每次進入下一輪迴圈時，`for` 都會重新把下一個值指定給 `i`。

因此即使在迴圈內寫：

```python
i = 3
```

也**不會改變 `range(3)` 下一次要提供的值**。

這段程式的輸出為：

```text
0
1
2
3
```

最後一輪執行 `i = 3` 後，迴圈結束，因此離開迴圈時 `i` 的值仍然是 `3`。

## 2. 函式參數與 `return`

下面定義一個加法函式 `add()`：

````python
def add(a, b):
    return a + b

a = 5
b = 10

print(add(a, b))
````

### 重點說明

函式：

```python
def add(a, b):
    return a + b
```

包含兩個參數：

- `a`
- `b`

`return` 會把運算結果傳回呼叫函式的位置。

因此：

```python
add(5, 10)
```

會回傳：

```text
15
```

再由 `print()` 把 `15` 顯示出來。

## 3. `while` 迴圈與結束條件

原始版本使用布林變數 `flag` 控制迴圈：

````python
flag = True

while flag:
    a = int(input())

    if a == 0:
        flag = False

    print(a)
````

### 執行流程

只要 `flag` 是 `True`，程式就會繼續要求輸入。

當：

```python
a == 0
```

時：

```python
flag = False
```

因此下一次檢查 `while flag` 時，迴圈就會結束。

要注意：`print(a)` 位於 `if` 外面，所以即使輸入 `0`，程式仍然會先印出：

```text
0
```

之後才離開迴圈。

### 另一種常見寫法：使用 `break`

若 `0` 只是用來表示「停止輸入」，也可以直接使用 `break`：

````python
while True:
    a = int(input())

    if a == 0:
        break

    print(a)
````

這個版本和原始版本有一個小差異：

- 原始版本會把最後輸入的 `0` 印出來。
- `break` 版本在遇到 `0` 時直接離開，因此不會印出 `0`。

應該使用哪一種，要依題目需求決定。

## 4. `print()` 的回傳值是 `None`

原始程式：

````python
# print() 會把內容輸出到畫面，
# 但它本身的回傳值是 None。

a = print("a")
````

執行：

```python
a = print("a")
```

時會發生兩件事：

1. `print("a")` 在畫面上輸出 `a`。
2. `print()` 本身回傳 `None`，因此變數 `a` 最後存的是 `None`。

可以再檢查：

````python
print(a)
````

因此完整結果會是：

```text
a
None
```

可以把它理解成：

```text
資料 → print() → 顯示到畫面
                 ↓
              回傳 None
```

所以「顯示內容」和「函式回傳值」是兩件不同的事情。

## 5. 自訂輸出函式 `log()`

原始程式定義了一個簡單的 `log()` 函式：

````python
def log(text):
    print("...." + text)
````

例如：

````python
log("程式開始")
log("資料處理完成")
````

輸出：

```text
....程式開始
....資料處理完成
```

因為 `log()` 裡面沒有寫 `return`，所以它和許多只負責「執行動作」的函式一樣，預設回傳：

```python
None
```

例如：

````python
result = log("測試")
print(result)
````

會看到：

```text
....測試
None
```

這也和前面的 `print()` 概念互相呼應：  
**函式可以執行某個動作，但不一定要回傳資料。**

## 6. 本次重點整理

- `for i in range(...)`：每輪的 `i` 由迭代器重新指定。
- 在 `for` 內手動修改 `i`，不會改變 `range()` 下一次產生的值。
- `return` 用來把函式的結果傳回去。
- `while` 可以使用布林變數控制，也可以搭配 `break`。
- `print()` 負責輸出內容，但回傳值是 `None`。
- 沒有明確寫 `return` 的函式，預設也會回傳 `None`。
