# 1. 輸入數值並依結果決定輸出格式

原始程式：

```python
r = input()
r = float(r)

R = (r * 9) / 5 + 32

if R.is_integer():
    print(int(R))
else:
    print(f"{R:.3f}")
```

由公式：

```text
R = r × 9 / 5 + 32
```

可以看出這是在做一種溫度換算形式；原始 Notebook 本身沒有寫出題目名稱，所以這裡重點放在 **數值轉型與輸出格式**。

程式流程：

```text
讀入資料
  ↓
input() 得到字串
  ↓
float() 轉成浮點數
  ↓
套用公式
  ↓
結果是整數值嗎？
 ┌───────────────┐
是               否
↓                ↓
轉成 int         保留 3 位小數
輸出             輸出
```


````python
r = input()
r = float(r)

R = (r * 9) / 5 + 32

if R.is_integer():
    print(int(R))
else:
    print(f"{R:.3f}")
````

## 1.1 為什麼 `input()` 後還要 `float()`？

Python 的：

```python
input()
```

**永遠先得到字串 `str`**。

例如輸入：

```text
20.5
```

實際上：

```python
r = input()
```

得到的是：

```python
"20.5"
```

不是數字 `20.5`。

所以若要做數學運算，需要：

```python
r = float(r)
```

也可以寫成一行：

```python
r = float(input())
```

---

### 常見錯誤

```python
r = input()
R = r * 9 / 5 + 32
```

這會出錯，因為 `r` 還是字串，不能直接做這組數值運算。


## 1.2 `is_integer()` 是什麼？

`float.is_integer()` 會檢查一個 `float` 是否代表整數值。

例如：

```python
3.0
```

雖然型別是 `float`，但數值沒有小數部分，所以：

```python
(3.0).is_integer()
```

得到：

```text
True
```

而：

```python
3.9
```

則得到：

```text
False
```


````python
f = 3.9

print(f.is_integer())
print(int(f))
````

### `int(f)` 不等於四捨五入

原始範例：

```python
f = 3.9
print(int(f))
```

輸出：

```text
3
```

`int()` 是把小數部分截掉，而不是四捨五入。

例如：

| 原值 | `int()` 結果 |
|---:|---:|
| `3.2` | `3` |
| `3.9` | `3` |
| `-3.9` | `-3` |

如果題目真的要求四捨五入，需要另外考慮 `round()` 等方法。


## 1.3 為什麼結果是整數時要再轉成 `int`？

假設：

```python
R = 68.0
```

直接：

```python

```

會顯示：

```text

```

但如果題目要求整數結果顯示：

```text
68
```

就可以：

```python
if R.is_integer():
    print(int(R))
```

這裡的 `int()` 主要是在控制輸出外觀。

# 3. f-string：控制小數位數

原始程式：

```python
f = 3.2

print(f"{f:2.2f}")
```

格式：

```python
f"{值:欄位寬度.小數位數f}"
```

因此：

```python
{f:2.2f}
```

可以拆成：

```text
f       → 要輸出的變數
2       → 最小欄位寬度
.2      → 小數點後 2 位
f       → 浮點數格式
```

但要注意：

> **欄位寬度是最小寬度，不是最大寬度。**

`3.20` 本身需要 4 個字元，因此即使寫寬度 `2`，Python 也不會把它截短。


````python
f = 3.2

print(f"{f:2.2f}")
print(f"{f:.2f}")
print(f"{f:8.2f}")
````

## 3.1 常用格式

### 固定兩位小數

```python
f"{x:.2f}"
```

例如：

```python
x = 5
```

得到：

```text
5.00
```

### 至少 8 格、兩位小數

```python
f"{x:8.2f}"
```

如果內容不足 8 格，預設會在左邊補空格。

### 右對齊

```python
f"{text:>8}"
```

### 左對齊

```python
f"{text:<8}"
```

### 置中

```python
f"{text:^8}"
```


# 4. `str.format()`：位置參數

原始程式：

```python
print("{0}{1}{0}".format("A", "B"))
```

其中：

```text
{0} → 第 0 個參數 → "A"
{1} → 第 1 個參數 → "B"
{0} → 再次使用第 0 個參數 → "A"
```

所以最後輸出：

```text
ABA
```

`{0}` 可以重複使用，這是這個例子的重點。


````python
print("{0}{1}{0}".format("A", "B"))
````

## 4.1 和 f-string 比較

`str.format()`：

```python
name = "Amy"
score = 90

print("{}: {}".format(name, score))
```

f-string：

```python
print(f"{name}: {score}")
```

兩者都可以。

在目前的 Python 中，f-string 通常更容易閱讀；但看懂 `.format()` 仍然很重要，因為舊程式或教材中很常出現。


# 5. `%`：餘數與循環問題

原始筆記包含：

```python
print(11 - 15)
print(-4 % 24)
print(3 % 2)
```

先看一般餘數：

```python
3 % 2
```

代表 3 除以 2 的餘數，因此結果：

```text
1
```

更有意思的是：

```python
-4 % 24
```

Python 的結果是：

```text
20
```

這讓 `% 24` 很適合處理 24 小時制的「繞回去」問題。


````python
print(11 - 15)
print(-4 % 24)
print(3 % 2)
````

## 5.1 為什麼 `% 24` 可以處理時間？

24 小時制可以想成一個循環：

```text
0, 1, 2, ..., 22, 23
↑                    ↓
└────────────────────┘
```

如果算出：

```text
-4
```

在 24 小時循環中，可以理解成從 0 往回 4 小時：

```text
20
```

Python：

```python
-4 % 24
```

正好得到：

```text
20
```

同樣地：

```python
25 % 24
```

得到：

```text
1
```

所以：

```python
某個時間 % 24
```

可以把結果限制回：

```text
0 ~ 23
```


## 5.2 原始輸入範例

原始程式：

```python
h = int(input())

print((h - 15) % 24)
```

它做的事情是：

1. 讀入整數 `h`
2. 減掉 `15`
3. 使用 `% 24` 把結果包回 `0 ~ 23`

例如：

```text
h = 11
```

先得到：

```text
11 - 15 = -4
```

再：

```text
-4 % 24 = 20
```

所以輸出：

```text
20
```

原始 Notebook 沒有保留完整題目敘述，因此這裡不額外指定它代表哪兩個時區，只說明程式本身的數學行為。


````python
h = int(input())

print((h - 15) % 24)
````

## 5.3 `%` 的另一個常見用途：判斷奇偶

```python
n % 2
```

如果：

```text
結果 = 0 → 偶數
結果 = 1 → 奇數
```

例如：

```python
8 % 2 == 0
```

所以 8 是偶數。

這也是 APCS / ZeroJudge 很常使用的基本技巧。


# 6. `//`：整數除法與分組

原始程式：

```python
n = int(input())

t = (n - 1) // 3
t += 1

print(t)
```

合併可以寫成：

```python
t = (n - 1) // 3 + 1
```

這個公式會把：

```text
1, 2, 3
```

分到第 1 組，

```text
4, 5, 6
```

分到第 2 組，

依此類推。


````python
n = int(input())

t = (n - 1) // 3
t += 1

print(t)
````

## 6.1 手動追蹤

| `n` | `n - 1` | `(n - 1) // 3` | 最後 `+ 1` | `t` |
|---:|---:|---:|---:|---:|
| 1 | 0 | 0 | 1 | 1 |
| 2 | 1 | 0 | 1 | 1 |
| 3 | 2 | 0 | 1 | 1 |
| 4 | 3 | 1 | 1 | 2 |
| 5 | 4 | 1 | 1 | 2 |
| 6 | 5 | 1 | 1 | 2 |
| 7 | 6 | 2 | 1 | 3 |

所以可以看到規律：

```text
每 3 個數字 → 同一組
```


## 6.2 為什麼先 `n - 1`？

如果直接：

```python
n // 3
```

那麼：

```text
1 // 3 = 0
2 // 3 = 0
3 // 3 = 1
```

`3` 已經跳到下一個結果，無法讓 `1, 2, 3` 都屬於同一組。

先減 1：

```python
(n - 1) // 3
```

就變成：

```text
1 → 0 // 3 → 0
2 → 1 // 3 → 0
3 → 2 // 3 → 0
```

最後再 `+1`，就得到第 1 組。

# 9. 常見錯誤整理

## 錯誤 1：函式要求回傳，卻只使用 `print()`

錯誤：

```python
def q2_two_decimal(x):
    print(f"{x:.2f}")
```

這會顯示資料，但函式真正的回傳值是：

```python
None
```

正確：

```python
def q2_two_decimal(x):
    return f"{x:.2f}"
```

---

## 錯誤 2：把 `.2f` 寫在大括號外

錯誤：

```python
f"{x}.2f"
```

正確：

```python
f"{x:.2f}"
```

---

## 錯誤 3：把欄位寬度當成最大長度

```python
f"{x:4}"
```

的 `4` 是：

```text
至少 4 格
```

不是「最多只能 4 格」。

---

## 錯誤 4：忘記輸出中的空格也是格式的一部分

以下兩個字串不同：

```text
Amy:90
Amy: 90
```

自動評測通常會要求輸出格式完全一致。

---

## 錯誤 5：需要循環時忘記 `%`

例如：

```text
-4 小時
```

若是 24 小時循環，不能直接當作合法小時輸出。

可以：

```python
-4 % 24
```

得到：

```text
20
```
