## 1. `input().split()`

如果使用者輸入：

```text
a b c
```

執行：

```python
input().split()
```

會得到一個 List：

```python
["a", "b", "c"]
```


````python
a = input().split()  # a b c
````

### 使用 Index 取得元素

List 的 index 從 `0` 開始。

因此：

```python
a[0]
a[1]
a[2]
```

會依序取得前三個元素。


````python
print(a)
# print(type(a))

print(a[0])
print(a[1])
print(a[2])
````

## 2. List Slicing

List slicing 可以取得 List 的一部分。

例如：

```python
lst[1:]
```

表示從 index `1` 開始，一直到最後。


````python
lst = ["a", "b", "c"]
lst_2 = lst[1:]

print(len(lst_2))
print(lst_2)
````

## 3. `print(*items, sep=...)`

在 `print()` 中使用：

```python
*lst
```

可以把 List 中的元素拆開，分別交給 `print()`。

而：

```python
sep=...
```

可以設定每個元素之間的分隔內容。


````python
lst = ["a", "b", "c", 1, "v", [1, 2, 3]]
lst_2 = lst[1:]

print(lst_2)
print(*lst_2, sep=lst[0])
````

### 使用 List 的第一個元素當連接文字

原始範例：

```python
["and", "apple", "banana", "cantaloupe"]
```

先把第一個元素 `"and"` 分開，再用它連接後面的水果名稱。


````python
in_1 = ["and", "apple", "banana", "cantaloupe"]
# in_1 = input().split()  <- and apple banana cantaloupe

in_2 = in_1[1:]

print(*in_2, sep=" " + in_1[0] + " ")
````

## 4. 字串比較

下面的原始例子先把整數 `a` 轉成字串，再進行字串比較：

```python
str(a) > "10"
```

這裡比較的是**字串**，不是整數大小。


````python
a = 2

# a > int("2")
str(a) > "10"  # "2" > "10"
````

## 5. `and` 與 `or`

Python 的 Boolean 運算中：

```text
and
```

通常會先於：

```text
or
```

計算。

下面保留原始 notebook 的幾個邏輯運算例子。


````python
# and -> or
print(True or False and True)  # True or False -> True
print(False and False or True)  # False or True -> True

print(
    True or False and True and False
    or True and False
    or True and True
)

print(
    1 == "1" or 1 >= 1 and False or True
)
````

## 6. 字串串接

兩個字串可以使用：

```python
+
```

連接。

如果中間需要空格，可以另外加入：

```python
" "
```


````python
s1 = "hello"
s2 = "hi"

print(s1 + " " + s2)  # hello hi
````

### Q3：班級分組

使用：

```python
//
```

取得完整班數，並使用：

```python
%
```

取得剩餘人數。


````python
# Q3
n, k = input().split()
n = int(n)
k = int(k)

c = n // k  # 20 // 3 -> 6
d = n % k   # 20 % 3 -> 2

print(f"可以分成{c}班，剩下{d}人。")
````
