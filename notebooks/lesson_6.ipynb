# 1. 格式化輸出

Python 有多種方式可以控制輸出的格式。

這裡比較三種常見寫法：

- `%` 格式化
- `.format()`
- f-string


````python
num = 6

# 1. % 格式化
print("%d" % num)
print("%f" % num)

print("---------")

# 2. .format()
print("{}".format(num))
print("{:.2f}".format(num))

print("---------")

# 3. f-string
print(f"{num}")
print(f"{num:.2f}")
````

## 控制小數位數

例如：

```python
f = 10
print("%.3f" % f)
```

`.3f` 代表：

> 以浮點數格式顯示，並保留 **3 位小數**。


````python
f = 10

print("%.3f" % f)
````

格式化之後的結果也可以存進變數。

```python
a = "%.3f" % f

```

此時 `a` 是一個**字串（string）**。


````python
a = "%.3f" % f

print(a)
print(type(a))
````

# 2. 條件判斷：`if / elif / else`

基本結構：

```python
if 條件:
    條件成立時執行
else:
    條件不成立時執行
```

如果有三種以上的情況，可以加入 `elif`：

```python
if 條件1:
    ...
elif 條件2:
    ...
else:
    ...
```


## 練習 1：判斷是否為正數

輸入一個整數 `x`：

- 如果 `x > 0`，印出 `"positive"`
- 否則印出 `"not positive"`

### Example

Input:

```text
5
```

Output:

```text
positive
```


````python
x = int(input())

if x > 0:
    print("positive")
else:
    print("not positive")
````

## 練習 2：判斷奇偶數

如果一個整數除以 2 的餘數是 0，就代表它是偶數。

```python
x % 2 == 0
```

其中 `%` 是**取餘數運算子**。

### Example

Input:

```text
7
```

Output:

```text
odd
```


````python
x = int(input())

if x % 2 == 0:
    print("even")
else:
    print("odd")
````

## 練習 3：判斷正數、負數或零

將輸入分成三種情況：

```text
x > 0   → positive
x < 0   → negative
其他    → zero
```

這時就需要使用 `elif`。

### Example

Input:

```text
0
```

Output:

```text
zero
```


````python
x = int(input())

if x > 0:
    print("positive")
elif x < 0:
    print("negative")
else:
    print("zero")
````

## 練習 4：判斷是否在範圍內

判斷 `x` 是否位於：

$$
1 \leq x \leq 10
$$

Python 可以直接寫成：

```python
1 <= x <= 10
```

### Example

Input:

```text
8
```

Output:

```text
in range
```


````python
x = int(input())

if 1 <= x <= 10:
    print("in range")
else:
    print("out of range")
````

## 練習 5：判斷登入狀態

已知正確帳號與密碼：

```text
username = admin
password = 1234
```

只有**帳號與密碼都正確**時，才印出：

```text
success
```

因此需要使用：

```python
and
```

代表「兩個條件都必須成立」。


````python
username = input()
password = input()

if username == "admin" and password == "1234":
    print("success")
else:
    print("fail")
````

# 3. `for` 與 `while`

可以先用一個簡單概念區分：

| 迴圈 | 常見使用情況 |
|---|---|
| `for` | 大致知道要重複幾次 |
| `while` | 不一定知道次數，只知道「什麼條件下要繼續」 |

例如：

```python
for i in range(10):
    print(i)
```

會執行固定次數。

但是登入、猜密碼等問題，通常不知道使用者需要嘗試幾次，因此很適合使用 `while`。


## `while True`

```python
while True:
    ...
```

代表：

> 一直重複執行。

如果沒有停止條件，就會形成**無限迴圈**。

因此常搭配：

```python
break
```

來離開迴圈。


## 範例：輸入正確關鍵字才停止

正確答案是：

```text
abc
```

只要輸入錯誤，就繼續要求輸入。

輸入正確後使用 `break` 結束迴圈。


````python
keyword_ans = "abc"

while True:
    keyword = input()

    if keyword == keyword_ans:
        print("success")
        break
    else:
        print("fail")
````

## 進階範例：帳號正確後只重新輸入密碼

假設正確資料為：

```text
帳號：Abc
密碼：1234
```

我們希望：

1. 帳號錯誤 → 帳號、密碼都重新輸入
2. 帳號正確但密碼錯誤 → 下一次只需要重新輸入密碼
3. 帳號、密碼都正確 → 顯示 `success` 並結束

這裡使用 `flag_account` 記錄：

> 帳號是否還需要重新輸入。


````python
account = "Abc"
password = "1234"

flag_account = True

while True:

    # 只有帳號尚未確認正確時，才重新輸入帳號
    if flag_account:
        username = input("Account: ")

    password_input = input("Password: ")

    if username == account and password_input == password:
        print("success")
        break

    elif username == account:
        # 帳號已經正確，下一輪不需要再輸入
        flag_account = False
        print("wrong password")

    else:
        print("fail")
````

# 重點整理

## 條件判斷

```python
if condition:
    ...
elif another_condition:
    ...
else:
    ...
```

## 常見條件運算

```python
x > 0
x < 0
x == 0
x != 0
1 <= x <= 10
```

## 邏輯運算

```python
condition1 and condition2
```

代表兩個條件都必須成立。

## `while`

```python
while condition:
    ...
```

當條件成立時持續執行。

## 無限迴圈 + `break`

```python
while True:
    ...
    if 結束條件:
        break
```

其中：

> `break` = **立刻離開目前的迴圈**。
