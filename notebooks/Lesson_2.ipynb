## 1. 使用 `split()` 拆開輸入

如果輸入資料使用逗號分隔，例如：

```text
A,B,C
```

可以使用：

```python
input().split(",")
```

將資料拆開，再分別存入 `a`、`b`、`c`。


````python
# a,b,c
a, b, c = input().split(",")

print(a, b, c)
````

### 指定其他分隔符號

`split()` 的括號中可以指定要使用的分隔符號。

例如：

```python
"a/b/c".split("/")
```

會按照 `/` 拆開字串。


````python
"a/b/c".split("/")
````

### 不同分隔格式

原始 notebook 另外記錄了幾種資料格式：

```text
A/D/E
2$5$3
4*5*0
5^p^v
```

如果要拆開這些資料，需要依照實際使用的符號指定 `split()` 的分隔符號。


````python
# A/D/E
# 2$5$3
# 4*5*0
# 5^p^v
````

## 2. 不要覆蓋內建函式名稱

`print` 是 Python 的內建函式。

原始程式提醒：

```python
# print = 5
```

如果把數值指定給 `print`，原本的 `print()` 功能就會被變數名稱取代。


````python
print("a")

# print = 5  # 原來的 print 功能被取代
````

## 3. `bool()` 與字串

下面的例子把一個非空字串：

```python
'""'
```

傳入 `bool()`。


````python
bool('""')
````

### 非空字串

即使字串內容看起來像兩個雙引號：

```python
'""'
```

它仍然是一個有內容的字串。

因此這個例子是在觀察「非空字串」轉成 Boolean 的結果。


## 4. 字串中的引號

如果外層使用雙引號：

```python
"..."
```

字串內容就可以直接包含單引號。


````python
print("'fnk'")
````

## 5. Q1：姓名與年齡

使用一行程式建立變數 `name` 和 `age`，分別存入你的名字與年齡，並輸出。

原始 notebook 的解法使用：

```python
input().split()
```

讓使用者在同一行輸入兩個資料。


````python
name, age = input().split()

print(name, age)
````

## 6. Q2：建立兩個變數並相加

使用一行程式建立：

```python
x = 8
y = 12
```

然後輸出：

```python
x + y
```


````python
x, y = 8, 12

print(x + y)
````

## 7. Q3：一行輸入兩個資料

讓使用者輸入兩個數字，例如：

```text
7 9
```

分別存入 `a`、`b`，並輸出兩個變數。


````python
a, b = input().split()

print(a, b)
````

## 8. Q4：轉成整數後相加

承接上一題。

`input().split()` 得到的資料是字串，因此先使用：

```python
int()
```

將 `a`、`b` 轉成整數，再計算：

```python
a + b
```


````python
a, b = input().split()
a, b = int(a), int(b)

print(a + b)
````

## 9. Q5：飲料名稱與價格

### 題目

使用者輸入：

```text
飲料名稱 價格
```

例如：

```text
milk_tea 60
```

輸出：

```text
milk_tea 的價格是 60 元
```

原始 notebook 使用 f-string 組合輸出文字。


````python
a, b = input("飲料名稱 價格").split()

# print(a, end=" ")
# print("的價格是 ", end="")
# print(b, end=" ")
# print("元")

print(f"{a} 的價格是 {b} 元")
````

## 10. 重點整理

### 空白分隔輸入

```python
a, b = input().split()
```

### 指定分隔符號

```python
a, b, c = input().split(",")
```

```python

```

### 一次指定多個變數

```python
x, y = 8, 12
```

### f-string

```python

```

### 避免使用內建函式名稱當變數

例如不要寫：

```python

```

否則會覆蓋原本的 `print()`。
